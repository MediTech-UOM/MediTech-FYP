```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt

# Load and preprocess data
file_path = "clinical_treatment_data_cleaned_V5.xlsx"
df = pd.read_excel(file_path)

def compute_reward(row):
    survival_reward = row['days_to_last_followup'] / 1000 if pd.notna(row['days_to_last_followup']) else 0
    vital_bonus = 1 if row['vital_status'] == 1 else -1
    recurrence_penalty = -1 if row.get('new_tumor_event_after_initial_treatment', 0) == 1 else 0
    response_score = 0
    response = str(row.get('measure_of_response', '')).lower()
    if 'complete' in response:
        response_score = 2
    elif 'partial' in response:
        response_score = 1
    elif 'stable' in response:
        response_score = 0.5
    elif 'progressive' in response:
        response_score = -1
    reward = survival_reward + vital_bonus + recurrence_penalty + response_score
    return reward

df['reward'] = df.apply(compute_reward, axis=1)

top_drug_list = ['cytoxan', 'tamoxifen', 'cyclophosphamide', 'paclitaxel', 'arimidex',
                 'doxorubicin', 'taxol', 'taxotere', 'anastrozole', 'adriamycin']
df_filtered = df[df['drug_name'].isin(top_drug_list)].copy()

le = LabelEncoder()
df_filtered['action'] = le.fit_transform(df_filtered['drug_name'])

features = [
    'age_at_initial_pathologic_diagnosis',
    'breast_carcinoma_estrogen_receptor_status',
    'breast_carcinoma_progesterone_receptor_status',
    'lab_proc_her2_neu_immunohistochemistry_receptor_status',
    'Basal-Like', 'HER2-Enriched', 'Luminal-A', 'Luminal-B'
]

df_filtered = df_filtered.dropna(subset=features + ['reward'])
X_state = df_filtered[features].astype(float).values
y_action = df_filtered['action'].values
y_reward = df_filtered['reward'].values

# Prepare data for model
X_train, X_test, y_train_act, y_test_act, y_train_rew, y_test_rew = train_test_split(
    X_state, y_action, y_reward, test_size=0.2, random_state=42
)
```
