```python
import plotly.graph_objects as go
import numpy as np

# Use spring layout in 3D
pos_3d = nx.spring_layout(G_sample, dim=3, seed=42)

# Extract node coordinates
x_nodes = [pos_3d[n][0] for n in G_sample.nodes()]
y_nodes = [pos_3d[n][1] for n in G_sample.nodes()]
z_nodes = [pos_3d[n][2] for n in G_sample.nodes()]

# Map node colors
color_map = {'author': 'skyblue', 'paper': 'lightgreen', 'field_of_study': 'lightpink'}
node_colors = [color_map[G_sample.nodes[n]['node_type']] for n in G_sample.nodes()]

# Create edge traces for each edge type
def make_edge_trace(edge_list, color):
    edge_x = []
    edge_y = []
    edge_z = []
    for edge in edge_list:
        x0, y0, z0 = pos_3d[edge[0]]
        x1, y1, z1 = pos_3d[edge[1]]
        edge_x += [x0, x1, None]
        edge_y += [y0, y1, None]
        edge_z += [z0, z1, None]
    return go.Scatter3d(
        x=edge_x, y=edge_y, z=edge_z,
        mode='lines',
        line=dict(color=color, width=2),
        hoverinfo='none'
    )

edges_writes = [(u, v) for u, v, d in G_sample.edges(data=True) if d['edge_type'] == 'writes']
edges_cites = [(u, v) for u, v, d in G_sample.edges(data=True) if d['edge_type'] == 'cites']
edges_hasTopic = [(u, v) for u, v, d in G_sample.edges(data=True) if d['edge_type'] == 'has_topic']

edge_trace_writes = make_edge_trace(edges_writes, 'blue')
edge_trace_cites = make_edge_trace(edges_cites, 'red')
edge_trace_hasTopic = make_edge_trace(edges_hasTopic, 'purple')

# Node trace
node_trace = go.Scatter3d(
    x=x_nodes, y=y_nodes, z=z_nodes,
    mode='markers+text',
    marker=dict(size=8, color=node_colors, line=dict(width=0.5, color='black')),
    text=[n for n in G_sample.nodes()],
    textposition="top center",
    hoverinfo='text'
)

fig = go.Figure(data=[edge_trace_writes, edge_trace_cites, edge_trace_hasTopic, node_trace])

fig.update_layout(
    title='Sampled Heterogeneous Graph (3D)',
    showlegend=True,
    legend=dict(itemsizing='constant'),
    margin=dict(l=0, r=0, b=0, t=40),
    width=1200, 
    height=580,
    scene=dict(
        xaxis=dict(showbackground=False),
        yaxis=dict(showbackground=False),
        zaxis=dict(showbackground=False),
    )
)

fig.show()
```

```python
import torch
import torch.nn.functional as F
from torch.nn import ModuleDict, Linear
from torch_geometric.nn import GATConv, HeteroConv

class HeteroGATLayer(torch.nn.Module):
    def __init__(self, metadata, hidden_dim, heads=1):
        super().__init__()
        self.convs = HeteroConv({
            edge_type: GATConv((-1, -1), hidden_dim, heads=heads, add_self_loops=False)
            for edge_type in metadata[1]
        }, aggr='sum')  # You can try 'mean', 'max', etc.

    def forward(self, x_dict, edge_index_dict):
        return self.convs(x_dict, edge_index_dict)
```

```python
class HAN(torch.nn.Module):
    def __init__(self, metadata, hidden_dim, out_dim, num_layers=2):
        super().__init__()
        self.layers = torch.nn.ModuleList()
        for _ in range(num_layers):
            self.layers.append(HeteroGATLayer(metadata, hidden_dim))

        # Final classifier for author nodes
        self.classifier = Linear(hidden_dim, out_dim)

    def forward(self, x_dict, edge_index_dict):
        for layer in self.layers:
            x_dict = layer(x_dict, edge_index_dict)
            x_dict = {key: F.relu(x) for key, x in x_dict.items()}

        return self.classifier(x_dict['author'])  # We're predicting for authors
```

```python
model = HAN(data.metadata(), hidden_dim=32, out_dim=2)  # Assuming binary classification
optimizer = torch.optim.Adam(model.parameters(), lr=0.005, weight_decay=5e-4)

for epoch in range(100):
    model.train()
    optimizer.zero_grad()
    out = model(data.x_dict, data.edge_index_dict)

    loss = F.cross_entropy(out, data['author']['y'].long())
    loss.backward()
    optimizer.step()

    print(f"Epoch {epoch}, Loss: {loss.item():.4f}")
```
